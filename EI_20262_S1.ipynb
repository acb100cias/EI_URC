```python
import numpy.random as rd
import numpy as np
import seaborn as sb
import scipy.stats as st
```

```python
sb.displot(rd.random(100000))
```

```python
sb.histplot(rd.random(100000), bins=100)
```

```python
sb.displot(rd.exponential(0.5,1000))
```

```python
sb.histplot(rd.exponential(0.5,1000), bins=100)
```

```python
st.skew(rd.random(100000))
```

```python
st.skew(rd.exponential(0.5,100000))
```

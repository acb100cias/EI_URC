```python
import numpy as np
import scipy.stats as st
```

```python
B1=st.bernoulli.rvs(p=0.2, size=10)
```

```python
alfa=0.05
confianza=1-alfa
```

```python
intervalo = st.t.interval(confianza, df=len(B1)-1, loc=np.mean(B1), scale=st.sem(B1))
print(intervalo)
```

```python
intervalo[1]-intervalo[0]
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
import scipy.optimize as opt
import math
```

```python
np.random.seed(42)
```

```python
l=5
muestra=np.random.poisson(l,100)
```

```python
muestra
```

```python
plt.hist(muestra)#,bins=35)
```

```python
def pois_log_vero(theta,muestra):
  log_vero=np.sum(muestra*np.log(theta)-theta-np.log([math.factorial(x) for x in muestra]))
  return log_vero
```

```python
L=lambda params:-pois_log_vero(params[0],muestra)
```

```python
emv=opt.minimize(L,x0=1)
```

```python
emv
```

```python
emv.x[0]
```

```python
pm=np.mean(muestra)
```

```python
pm
```

```python
lvals=np.linspace(0.1,10,100)
```

```python
lvals
```

```python
vero=[pois_log_vero(lam,muestra) for lam in lvals]
```

```python
vero
```

```python
from matplotlib import lines
plt.figure(figsize=(10,6))
plt.plot(lvals,vero)
plt.axvline(emv.x[0],color='r',linestyle='--')
plt.scatter([emv.x[0]],[pois_log_vero(emv.x[0],muestra)],color='r')
plt.scatter([l],[pois_log_vero(l,muestra)],color='g',marker='^')
```
